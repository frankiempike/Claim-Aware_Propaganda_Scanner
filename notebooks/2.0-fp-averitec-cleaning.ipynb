```python
import pandas as pd
import os
from pathlib import Path
```

```python
#Set up paths
BASE_DIR = Path("..").resolve()
INTERIM_DIR = BASE_DIR / "data" / "interim"
PROCESSED_DIR = BASE_DIR / "data" / "processed"
os.makedirs(PROCESSED_DIR, exist_ok=True)
```

```python
#Load AveriTec data
df_av = pd.read_csv(INTERIM_DIR / "averitec_dev_flattened.csv")
df_av.head()
```

```python
#Handle missing values
df_av['speaker'] = df_av['speaker'].fillna('Unknown')
df_av['claim_date'] = df_av['claim_date'].fillna('Unknown Date')
df_av.head()
```

```python
#Convert labels to consistent format that is less wordy
label_mapping = {
    'Supported': 'true',
    'Refuted': 'false',
    'Not Enough Evidence': 'unverifiable',
    'Conflicting Evidence': 'mixed'
}

df_av['verdict'] = df_av['label'].map(label_mapping)
df_av.head()
```

```python
#Save to processed data folder
output_path = BASE_DIR / "data" / "processed" / "averitec_cleaned.csv"
df_av.to_csv(output_path, index=False)
print("AVeriTeC DF is cleaned")
```
