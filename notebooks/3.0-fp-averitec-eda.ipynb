```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
#This notebook is for EDA of AveriTec data
BASE_DIR = Path("..").resolve()
df_av = pd.read_csv(BASE_DIR / "data" / "interim" / "averitec_dev_flattened.csv")
```

```python
print(f"Total claims: {len(df_av)}")
print(df_av['label'].value_counts())
```

```python
#Visualize labels
plt.figure(figsize=(5, 3))
sns.countplot(data=df_av, y='label', order=df_av['label'].value_counts().index)
plt.title('Distribution of Fact-Checking Verdicts in AVeriTeC')
plt.xlabel('Count')
plt.ylabel('Verdict')
plt.show()
```

```python
#Visualize common speakers
plt.figure(figsize=(5, 3))
df_av['speaker'].value_counts().head(10).plot(kind='barh')
plt.title('Top 10 Most Frequent Claim Speakers')
plt.gca().invert_yaxis()
plt.show()
```
